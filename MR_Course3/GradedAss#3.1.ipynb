```python
#------- ALL NECESSARY LIBRARIES -------#
import numpy as np
import math as mt
import modern_robotics as mr
```

```python
#---- PROBLEM # 1 -------

"""ENCAPSULATED FUNCTIONS FOR PROBLEMS1"""
def find_mass(volume, density):
    return volume*density 

def find_cylinder_inertia_zz(mass, radius, length):
    Ixx = Iyy = (1/12)*mass*(3*mt.pow(radius,2)+mt.pow(length,2))
    Izz = 1/2*mass*mt.pow(radius,2)
    
    return np.array([Ixx, Iyy, Izz])

def find_sphere_inertia(mass, radius):
    Ixx = Iyy = Izz = (2/5)*mass*mt.pow(radius,2)
    
    return np.array([Ixx, Iyy, Izz])

def apply_parallel_axis_z_shift(I_base, mass, z_distance):
    shift = mass * mt.pow(z_distance, 2)
    I_base[0] = I_base[0] + shift
    I_base[1] = I_base[1] + shift
    
    return I_base
```

```python
"""PROBLEM1 EXECUTION BLOCK"""
# all in terms meters
density = 5600
r_cylinder = 0.02
L_cylinder = 0.2
r_sphere = 0.1
z_distance = (L_cylinder/2)+r_sphere

volume_sphere = (4/3)*mt.pi*mt.pow(r_sphere, 3)
volume_cylinder = mt.pi*mt.pow(r_cylinder, 2)*L_cylinder

mass_cylinder = find_mass(volume_cylinder, density)
mass_sphere = find_mass(volume_sphere, density)

I_base_cylinder = find_cylinder_inertia_zz(mass_cylinder, r_cylinder, L_cylinder)
I_base_sphere = find_sphere_inertia(mass_sphere, r_sphere)

I_shiftedSphere = apply_parallel_axis_z_shift(I_base_sphere, mass_sphere, z_distance)

np.diag(I_base_cylinder+2*I_shiftedSphere)
```
