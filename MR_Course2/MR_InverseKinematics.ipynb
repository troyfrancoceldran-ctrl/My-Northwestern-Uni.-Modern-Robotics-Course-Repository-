```python
def manual_newton_raphson(init_guess, iterations):
    # Set the initial guess
    vars = np.array(init_guess, dtype=float)
    
    for i in range(iterations):
        x, y = vars
        
        # 1. Evaluate the equations F(x, y)
        F = np.array([x**2 - 9, y**2 - 4])
        
        # 2. Evaluate the Jacobian matrix
        J = np.array([
            [2*x, 0],
            [0, 2*y]
        ])
        
        # 3. Calculate the step using linear algebra (J * delta = -F)
        delta = np.linalg.solve(J, -F)
        
        # 4. Update the variables
        vars = vars + delta
        
    return vars

# Run for exactly 2 iterations
solution = manual_newton_raphson([1.0, 1.0], 2)

print(f"x = {solution[0]:.2f}")
print(f"y = {solution[1]:.2f}")
```
